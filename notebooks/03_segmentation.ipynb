# 03. Product Segmentation and Layout Analysis
Demonstrates Otsu thresholding, morphological opening/closing, bounding box, and centroid calculation.


```python
import os, sys
sys.path.append("..")
import matplotlib.pyplot as plt
from src.preprocessing import preprocess_pipeline
from src.segmentation import segment_product
from src.object_analysis import analyze_object_and_background
```

```python
sample_img_path = os.path.join("..", "dataset", "test", "good", os.listdir("../dataset/test/good")[0])
prep = preprocess_pipeline(sample_img_path)
rgb = prep["resized_rgb"]
gray = prep["grayscale"]

seg = segment_product(rgb, gray)
obj = analyze_object_and_background(rgb, gray, seg["binary_mask"])

print("Reliability:", seg["reliability_status"])
print("Area Ratio: ", seg["area_ratio"])
print("Centroid:   ", obj["centroid"])
print("Alignment:  ", obj["alignment_score"])
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
axes[0].imshow(rgb); axes[0].set_title("Input RGB")
axes[1].imshow(seg["binary_mask"], cmap="gray"); axes[1].set_title("Morphological Mask")
axes[2].imshow(seg["white_bg_rgb"]); axes[2].set_title("White Studio BG")
axes[3].imshow(obj["annotated_image"]); axes[3].set_title("BBox and Centroid")
for ax in axes: ax.axis("off")
plt.tight_layout()
plt.show()
```
