# 04. Classification and Benchmarking
Demonstrates loading features, evaluating baseline rules, and testing the Random Forest model.


```python
import os, sys
sys.path.append("..")
import pandas as pd
from src.classifier import load_model, evaluate_classifier
```

```python
df = pd.read_csv("../dataset/extracted_features.csv")
train_df = df[df["split"] == "train"]
test_df = df[df["split"] == "test"]
print(f"Loaded {len(train_df)} train items and {len(test_df)} test items.")
df.head()
```

```python
model = load_model("../models/random_forest.joblib")
eval_res = evaluate_classifier(model, test_df, test_df["ground_truth"])
print("Accuracy: ", eval_res["accuracy"])
print("Precision:", eval_res["precision"])
print("Recall:   ", eval_res["recall"])
print("F1 Score: ", eval_res["f1_score"])
print("Feature Importances:", eval_res["feature_importances"])
```
