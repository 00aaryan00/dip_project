```python
import os, sys
sys.path.append("..")
import cv2
import numpy as np
import matplotlib.pyplot as plt
from src.preprocessing import preprocess_pipeline
```

```python
sample_img_path = os.path.join("..", "dataset", "test", "good", os.listdir("../dataset/test/good")[0])
prep = preprocess_pipeline(sample_img_path, max_dimension=500)

print("Original shape:", prep["original_shape"])
print("Processed shape:", prep["processed_shape"])
```
