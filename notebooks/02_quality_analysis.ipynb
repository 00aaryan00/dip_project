# 02. Quality Analysis, Enhancement, and Filtering
Demonstrates brightness, contrast, sharpness, noise estimation, and spatial filters.


```python
import os, sys
sys.path.append("..")
import matplotlib.pyplot as plt
from src.preprocessing import preprocess_pipeline
from src.quality_analysis import analyze_brightness_contrast
from src.sharpness import analyze_sharpness
from src.noise import analyze_noise
```

```python
sample_img_path = os.path.join("..", "dataset", "test", "good", os.listdir("../dataset/test/good")[0])
prep = preprocess_pipeline(sample_img_path)
gray = prep["grayscale"]

qc = analyze_brightness_contrast(gray)
sharp = analyze_sharpness(gray)
noise = analyze_noise(gray)

print("Brightness:", qc["brightness_score"], qc["brightness_status"])
print("Contrast:  ", qc["contrast_score"], qc["contrast_status"])
print("Sharpness: ", sharp["sharpness_score"], sharp["sharpness_status"])
print("Noise:     ", noise["noise_score"], noise["noise_status"])
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].imshow(gray, cmap="gray")
axes[0, 0].set_title("Original Grayscale")
axes[0, 1].imshow(qc["hist_equalized"], cmap="gray")
axes[0, 1].set_title("Histogram Equalized")
axes[1, 0].imshow(sharp["laplacian_display"], cmap="gray")
axes[1, 0].set_title("Laplacian Edges")
axes[1, 1].imshow(noise["median_filtered"], cmap="gray")
axes[1, 1].set_title("Median Filtered")
for ax in axes.ravel(): ax.axis("off")
plt.tight_layout()
plt.show()
```
